##**Problema de negocio**

Se trata de un algoritmo capaz de analizar las caractericticas de una popiedad -como comodidades, tamaño, ocupacion del espacio en un periodo determinado- y sugerir al anfitrion un precio a cobrar por tarifas diarias que garantice ganancias en momentos de alta demanda.

###**1. Entendiendo el problema**

```python
import pandas as pd
```

```python
datos_hosting = pd.read_json('datos_hosting.json')
```

```python
datos_hosting.head()
```

```python
datos_hosting = pd.json_normalize(datos_hosting['info_inmuebles'])
datos_hosting.head()
```

```python
datos_hosting.info()
```

###**2. Datos numericos**

```python
columnas = list(datos_hosting.columns)
columnas
```

```python
datos_hosting = datos_hosting.explode(columnas[3:])
datos_hosting
```

```python
datos_hosting.reset_index(inplace=True, drop=True)
datos_hosting.head()
```

```python
datos_hosting.info()
```

```python
import numpy as np
```

```python
datos_hosting['max_hospedes'] = datos_hosting['max_hospedes'].astype(np.int64)
```

```python
col_num = ['cantidad_baños','cantidad_cuartos','cantidad_camas']
```

```python
datos_hosting[col_num] = datos_hosting[col_num].astype(np.int64)
datos_hosting.info()
```

```python
datos_hosting['evaluacion_general'] = datos_hosting['evaluacion_general'].astype(np.float64)
```

```python
datos_hosting.info()
```

```python
datos_hosting['precio'] = datos_hosting['precio'].apply(lambda x: x.replace('$','').replace(',','').strip())
```

```python
datos_hosting['precio'] = datos_hosting['precio'].astype(np.float64)
```

```python
datos_hosting.info()
```

```python
datos_hosting[['cuota_deposito','cuota_limpieza']] = datos_hosting[['cuota_deposito','cuota_limpieza']].applymap(lambda x: x.replace('$','').replace(',','').strip())
```

```python
 datos_hosting[['cuota_deposito','cuota_limpieza']] =  datos_hosting[['cuota_deposito','cuota_limpieza']].astype(np.float64)
```

```python
datos_hosting.info()
```

###**3. Datos de texto**

```python
datos_hosting.head()
```

```python
datos_hosting['descripcion_local'] = datos_hosting['descripcion_local'].str.lower()
```

```python
datos_hosting['descripcion_local'][3169]
```

```python
datos_hosting['descripcion_local'] = datos_hosting['descripcion_local'].str.replace('[^a-zA-Z0-9\-\']',' ', regex=True)
```

```python
datos_hosting['descripcion_local'] = datos_hosting['descripcion_local'].str.replace('(?<!\w)-(?!\w)',' ', regex=True)
```

```python
datos_hosting['descripcion_local'].head()
```

```python
datos_hosting['descripcion_local'] = datos_hosting['descripcion_local'].str.split()
```

```python
datos_hosting.head()
```

```python
datos_hosting['comodidades'] = datos_hosting['comodidades'].str.replace('\{|}|\"', '', regex=True)
```

```python
datos_hosting['comodidades'] = datos_hosting['comodidades'].str.split(',')
```

```python
datos_hosting.head()
```
